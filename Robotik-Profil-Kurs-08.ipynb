### Der Code

Hier unten ist der Code für den Calliope, der den Roboter etwa 10cm fahren lässt.

```python
##
#
# Zustände als Variablen
#
##
z0_start_ = 0
z1_fahren_ = 1
z2_stopp_ = 2
aktueller_zustand_ = z0_start_


##
#
# Initialisierung abschließen und abspielen eines Tons
#
##
basic.pause(5000)
music.play_tone(262, music.beat(BeatFraction.WHOLE))


##
#
# Übergangsfunktion in der on_forever Funktion
#
##
def on_forever():

    global aktueller_zustand_

    if aktueller_zustand_ == z0_start_:           # setzen des aktuellen Zustandes
        maqueen.motor_stop(maqueen.Motors.ALL)    # alle Motoren Stopp
        basic.show_number(aktueller_zustand_)     # anzeigen des aktuellen Zustandes 
        if input.button_is_pressed(Button.A):     # warten auf das drücken von Knopf A
            aktueller_zustand_ = z1_fahren_       # Wechsel des Zustandes

    if aktueller_zustand_ == z1_fahren_:          # Zustand fahren
        basic.show_number(aktueller_zustand_)     # anzeigen des Zustandes
        maqueen.motor_run(maqueen.Motors.ALL,     # ansprechen aller Motoren 
                          maqueen.Dir.CW,         # festlegen der Richtung
                          60)                     # festlegen der Schnelligkeit, Staerke
        basic.pause(60)                           # eine Pause für Anzahl Millisekunden
        aktueller_zustand_ = z2_stopp_            # Wechsel des Zustandes

    if aktueller_zustand_ == z2_stopp_:           # Zustand zwei
        basic.show_number(aktueller_zustand_)     # anzeigen des Zustandes
        maqueen.motor_stop(maqueen.Motors.ALL)    # Motoren Stopp
        basic.pause(500)                          # Pause von 500 ms
        aktueller_zustand_ = z0_start_            # gehe zurück zu Start
        
basic.forever(on_forever)
```
