## Story

Der Roboter **Bob** ist auf einer Mission auf einem fremden Planeten, wohin genau ist streng geheim.

<img src="../shared/img_curie_course/Robo_Bob.png" alt="Bob the Robot" style="width: 450px;"/>

Abb. 1: Bob benötigt deine Hilfe

Bob ist auf einem fremden Planeten namens **Blub**, auf Blub sprechen alle **“Ziserb”**, das ist eine echt komische Sprache. Diese Sprache kann dir in Teilen bekannt vorkommen.

Es gibt ein Alphabet und dieses Alphabet wird **Zis** genannt und besteht aus zwei Zeichen:
- a
- b

Das heißt, es ist möglich aus diesen beiden Zeichen, Wörter der Sprache **“Ziserb”** zu bilden.

Diese Worte sehen wie folgt aus:

- ab
- aaaaaab
- b
- aaaab
- aaaaaaaaaaaaaaaaaaaaaaaaaab

**Was fällt dir auf?**

**Aufgabe 2**:
Formuliere eine Regel bzw. einen Satz, der die Sprache **“Ziserb”** beschreibt.

Schließlich gibt es auf diesem Planeten auch **andere Außerirdische**, sie sprechen auch wilde Sprachen, u. a. **Woki und Deutsch**.

Folgende Wörter sind nicht Teil der Sprache **“Ziserb”**:
- abc
- Klingel
- tingo
- bubbba
- ba
- aba
- bubble

Bob hat ein sehr intelligentes Sprachmodul zum Lernen der Sprache **“Ziserb”** und er besitzt ein ausgezeichnetes **Speech-to-Text** Übersetzungsmodul. Er kann somit **problemlos** gesprochene **Sprache in Text** umwandeln.

Leider ist beim letzten Update seiner Firmware sein **Wort-Filtermodul durcheinander** beschädigt worden. 

Das war der neue Mitarbeiter namens Olli welcher den Quellcode des “Ziserb” **Token Parser gelöscht** hat. 

Ein **Token Parser** ist ein kleines Programm, das auftretende **Zeichen in einer Zeichenkette einliest und verarbeitet**.

Um das Wort-Filtermodul zu reparieren, musst du einen Tokenparser als Zustandsautomat definieren.

## Bedingungen

Anbei gibt es noch einige Bedingungen bzw. Verabredungen:
- Es muss **mindestens einen Zustand** für das **Akzeptieren** eines Wortes geben.
- Es muss **mindestens einen Zustand** für das **Ablehnen** eines Wortes geben.
- Zustände sollen wie folgt benannt werden (**Z0** **bis Z[n]**), wobei **[n]** für eine beliebige positive, ganze Zahl steht
- Das Alphabet der Sprache **“Ziserb”** ist {a, b}.
- Die Gesamtheit aller <b>anderen Alphabete wird als Z*</b> bezeichnet und die **Zeichen** können einfach als * dargestellt werden.
- Ein Wort der Sprache **Ziserb** besteht mindestens aus einem Zeichen.

## Der Automat der **Ziserb** Sprache als grafisches Modell

Der Automat kann auch als grafisches Modell dargestellt werden.

Dabei ist der **Startzustand** als **Z0** definiert, mögliche **Endzustände** sind **schwarz und umrandet** dargestellt.

Hierbei gehen von einem Zustand, alle möglichen Ereignisse ab (d. h. "a" wird gelesen und "b" wird gelesen).

Wenn nichts gelesen wird, muss man nichts darstellen.
