## Zweck
Das Jupyter Notebook besteht aus Zellen, diese können ausgeführt werden. Es gibt Text-Zellen (Markdown) oder Code-Zellen.

## Ausführung der Zellen
An der Seite einer aktiven Zelle ist eine Markierung, durch drücken der Tasten Shift und Enter wird eine Zelle ausgeführt.

### Ausführen von Text-Zellen
Beim ausführen von Text-Zellen passiert nicht so viel, die Markierung am linken Rand springt nur zur nächsaten aktiven Zelle.

Durch Doppel-Klicken einer Text-Zelle gelangst du in den Editier-Modus und kannst die Zelle bzw. deren Text verändern.

### Ausführen von Code-Zellen

Bei der Ausführung von Code-Zellen wird der Code des geschriebenen Programms ausgeführt.
