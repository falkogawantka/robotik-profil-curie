### Minimum-Wissen: Zustand
Ein **Zustand** beschreibt eine **Phase** (**Modus** oder schlicht Zustand) in dem ein **System** (ein technisches System oder Software) **operiert**.

### Minimum-Wissen: Zustandsvariablen des Systems

Die **Zustandsvariablen des Systems** werden durch eine Menge von **Variablen** beschrieben, hierbei können gewisse Variablen einen **Zustand abbilden**, z.B.: den Startzustand.

### Minimum-Wissen: Übergang von Systemzuständen

Wenn nun ein äußerer **Reiz** (**ein Ergeinis**, engl. event) eintritt kann bei bestimmten Umständen (also Bedingungen, if -- if-else -- if-elif-else) ein **Zustandswechsel eingeleitet werden**.

### Minimum-Wissen: Akzeptor

Ein abstrakter **Akzeptor** ist ein Automat der für eine definierte **Eingabe** die Ausgabe **akzeptiert liefert oder nicht akzeptiert**, falls die Eingabe gegen Regeln des Akzeptors verstößt. 

Die Analogie war, dass wie wenn man einen Kuchen bäckt (dessen Rezept aus mehreren Schritten besteht) eine sich wiederkehrende Folge, wie das Abmessen einzelner Zutaten, als ein kleiner Hilfsprozess betrachtet wird (d.h. Funktion).

```python
##
# 5. Übergänge zwischen den states. --> Bitte ausführen, aber erst programmieren
#
# Dieser Code muss kopiert werden. --> kopieren in MakeCode aber erst später
##

def next_state(next_item_in_list_):             # Funktion die den Folgezustand berechnet für eine Eingabe
    
    global current_state_                       # Einlesen der vorher definierten Variable

    if (next_item_in_list_ == -1):              # Wenn das Zeichen unseres Wortes -1 ist dann 
        current_state_ = States.start           # gehe in den Startzustand --> ab hier hilft uns der 

    elif (current_state_ == States.start):      # Wenn wir im Zustand Start sind 

        if (next_item_in_list_ == 0):           # Wenn wir das Zeichen "a" lesen
            pass                                # dann soll was passieren ...

    elif (current_state_ == States.end):
        pass

    elif (current_state_ == States.trap):
        pass
```

```python
##
# Definition der on_forever Funktion, welche dauerhaft ausgeführt werden soll. --> Bitte ausführen!
#
# Dieser Code muss kopiert werden. --> kopieren in MakeCode aber erst später
##

def on_forever():
    global list_of_numbers_
    if input_sim.button_is_pressed(Button.A):
        list_of_numbers_.append(0) # append can be push 0 in MakeCode, can be "a" 
        for i in range(1, len(list_of_numbers_)):
            basic.clear_screen()
            basic.pause(400)
            basic.show_number(list_of_numbers_[i])
            
        basic.clear_screen()

        for i in range(0, len(list_of_numbers_)):
            next_state(list_of_numbers_[i])

        if current_state_ == States.end:
            basic.show_string("Accepted")
        else:
            basic.show_string("Rejected")
        
    elif input_sim.button_is_pressed(Button.B):
        list_of_numbers_.append(1) # append can be push in MakeCode, 1 can be "b"
        for i in range(1, len(list_of_numbers_)):
            basic.clear_screen()
            basic.pause(400)
            basic.show_number(list_of_numbers_[i])
            
        basic.clear_screen()    
            
        for i in range(0, len(list_of_numbers_)):
            next_state(list_of_numbers_[i])

        if current_state_ == States.end:
            basic.show_string("Accepted")
        else:
            basic.show_string("Rejected")
```

```python
##
# Ausführen der on_forever Funktion. --> Bitte ausführen!
# 
##

restart() # Diese Zeile muss nicht kopiert werden.

basic.forever_test(on_forever) # --> Nur diese Zeile kopieren in MakeCode.
```

# Beispiel Nummer 2 -- Eine Bedarfs-Ampel

Analog dazu entwickeln wir das Beispiel einer dreiphasigen Fußgänger Ampel.

```python
##
#
# 1. Zustandsvariablen des Systems
#
##

button_was_pressed_ = -1 # Knopf an der Ampel
state_ = -1              # allgemeiner Zustand des Automaten
```

```python
##
#
# 2. Zustände der Ampel
#
##

class State:             # Aufzählung aller Zustände
    start = -1
    red = 0
    green = 1
```

# 2. Aufgabe zur Bewertung:

Diese Aufgabe besteht darin die unten stehende Funktion zu erweitern, 

hierbei soll die Arbeitsweise einer Fußgänger Ampel nachgebildet werden.

Das heißt im Normalzustand ist die Ampel rot 

und bei drücken eines Knopfes wechselt sie auf grün.

Erweitere die untenstehende Funktion geeignet.

Das anzeigen der "0" stellt dabei die Rot-Phase dar.

Das anzeigen der Grün-Phase wird durch das darstellen einer "1" symbolisiert.

**Bonus:**

Es wird einen Bonus geben, 

wenn ihr die LEDs ansteuern könnt,

das heißt wenn die Ampel im Zustand rot (d.h. 0 ist) soll die LED rot leuchten.

Das gleiche soll analog bei grün passieren, auch mit der LED.

## Das Diagramm der Ampel

<img src="images/ampel_fsm_fussganger.png" alt="Unser Zustandsautomat" style="width: 650px;"/>

```python
##
#
# 3. Eine Funktion die den Folgezustand berechnet.
#
##

def next_state(button_was_pressed_):
    global state_

    if state_ == State.start:
        state_ = State.red
        basic.show_string("-")
    
    elif (state_ == State.red) and (button_was_pressed_ != 1):
        basic.show_string("0")

    elif (state_ == State.green):
        basic.show_string("1")
```

```python
##
#
# 4. Aufruf der Funktion zum Bestimmen der Zustände
#
##

def on_forever():
    global button_was_pressed_
    next_state(button_was_pressed_)
    
basic.forever_test(on_forever)
```
