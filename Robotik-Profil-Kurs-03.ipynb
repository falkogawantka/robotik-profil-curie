# Definitionen

An dieser Stelle werden die Definitionen eingeführt, auf welche am Anfang verzichtete wurde.

- **Zustand**: Beschreibt, was das System gerade ist oder tut  
  (z. B. „hungrig“, „misst Abstand“, „fährt hoch“).

- **Ereignis (Input)**: Ein Auslöser, der auftreten kann und ggf. einen Übergang anstößt  
  (z. B. „1‑€‑Münze eingeworfen“, „Taster gedrückt“, „Timer abgelaufen“).

- **Bedingung (Guard)**: Die Voraussetzung, unter der ein Ereignis tatsächlich zu einer Änderung führt  
  (z. B. „Betrag ≥ 1 €“, „Abstand < 10 cm“). Sie ist ein wahr/falsch‑Kriterium.

- **Aktion (Effekt)**: Das, was beim Übergang (oder beim Eintritt/Austritt in einen Zustand) ausgeführt wird.  
  Aktionen können interne Variablen verändern oder Ausgaben auslösen (z. B. LED an, Münze zurückgeben).

- **Übergang**: Der Wechsel von einem Zustand in einen anderen.  
  Er wird durch **Ereignis** + optional **Bedingung** ausgelöst und kann eine **Aktion** ausführen.

**Kurzschreibweise für einen Übergang:**  
`Ereignis [Bedingung] / Aktion : Zustand_A → Zustand_B`

## Ein Spindschloss im Schwimmbad - Beispiel 3

<img src='../shared/img_curie_course/euro_lock.jpeg' alt="FSM eat apple 2" style="width: 220px;">

Wir stellen uns ein Schloss ähnlich diesem vor. Mn muss das Schluss durch Münzeinwurf aktivieren und kann es mit dem Abschließen verriegeln.

Dieses Schloss hat folgende Zustände:
- Start-Zustand
- **Geschlossen**
- **Offen**

Die ausführbaren Aktionen sind:
- abschließen (bzw. Drehen des Schlüssel im Schloss)
- Münzeinwurf (d. h. es werden nur 1€ Münzen zugelassen)
- und der automatische Wechsel von Start --> abgeschlossen

<img src='../shared/img_curie_course/euro_lock_02.svg' alt="FSM eat apple 2" style="width: 680px;">

**Verhaltensbeschreibung des Automaten**

1. Wenn das elektronische Münzschloss angeht, wechselt es automatisch von Start (kleiner schwarzer Punkt) in den Zustand (**geschlossen**)

2. Hier kann man das Schloss verriegeln  (d. h. **abschließen**), aber das ändert nicht den Zustand des Schlosses.

3. **Wenn** man eine **1 Euro Münze einwirft**, **dann** **wechselt der Zustand** in **offen** und die Münze wird im Schloss geparkt (sie verbleibt im Schloss).

4. Ein weiterer Münzeinwurf führt zu einer Rückgabe der gerade eingeworfenen Münze, aber zu keiner Änderung des Zustandes.

5. Wenn der Zustand **offen** aktiv ist und eine Münze im Schloss steckt und man jetzt den Schlüssel umdreht (d. h. abschließt), dann verriegelt sich das Schloss und der Zustand wechselt zu **geschlossen**. Anschließend wird die Münze zurück gegeben.
