# While-Schleife:

Eine While-Schleife wiederholt so lange den Code-Block bis die festgelegte Bedingung nicht mehr erfüllt ist.

Eine While-Schleife benötigt eine Bedingung, um ausgeführt zu werden.
Du kannst diese Bedingung auch zu True oder False setzen. Eine Schleife mit der Bedingung True wird sich endlos wiederholen, während eine Schleife mit der Bedingung False sich nie wiederholt.

Wir werden für unsere Schleifen aber eher festlegen das sie sich solange wiederholt bis z.B. die Variable „nummer“ genauso hoch wie 5 ist.

Beispiel:

```python
nummer = 0
while nummer < 5:
    print("Nummer ist gerade:", nummer)
    nummer += 1

print("Schleife beendet.")
```

Eine Aufgabe aus dem ersten Kurs, war die Berechnung des MittelPunktes der **Seite a**, sprich die **Strecke zwischen Punkt A und Punkt B**

Bestimmt haben sich viele gedacht:

"Das ist einfach, da muss da muss ich einfach die x-Koordinate von Punkt B halbieren"

**Frage**: Ist das wirklich so einfach? 

Was ist wenn der Punkt A auf der x-Achse liegt, aber nicht am Koordinaten Ursprung sondern kleiner als 0 ist? 

Wie müsst du den Mittelpunkt dann berechnen?

**Antwort**: 

Der Mittelpunkt muss dann $\frac{{A}_x + B_x}{2} = $

**Frage**: Was ist wenn die x-Koordinate größer als 0 ist, funktioniert es dann auch so?

Gegeben ist ein Beispiel, das auf verschiedene Bedingungen prüft.

**Aufgabe 13b**:

vervollständige die Code-Abschnitte und die Berechnungen, wenn x von Punkt A größer bzw. kleiner als 0 ist.

Teste anschließend mit verschiedenen Werten manuell nach.

```python
##
# Code-Beispiel
# initiales zuweisen der Variablen
##

x_von_Punkt_A = 0
x_von_Punkt_B = 39
haelfte_von_seite_a = -1
```

```python
##
# Berechnen des Mittelpunktes wenn x von Punkt A: 0 ist
##

if x_von_Punkt_A == 0:
    haelfte_von_seite_a = x_von_Punkt_B/2
```

```python
##
# Berechnen des Mittelpunktes wenn x von Punkt A: größer 0 ist
##

if x_von_Punkt_A > 0:
    haelfte_von_seite_a = (x_von_Punkt_B - x_von_Punkt_A) / 2
```

```python
##
# Berechnen des Mittelpunktes wenn x von Punkt A: kleiner 0 ist
##

if x_von_Punkt_A < 0:
    haelfte_von_seite_a = (x_von_Punkt_B + abs(x_von_Punkt_A)) / 2 # absoluten Betrag von x von Punkt a nehmen
```

```python
##
# Ergebnis
##

haelfte_von_seite_a
```
